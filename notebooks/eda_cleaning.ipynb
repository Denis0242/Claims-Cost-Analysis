```
df["claim_date"] = pd.to_datetime(df["claim_date"], errors="coerce")
df["claim_amount"] = pd.to_numeric(df["claim_amount"], errors="coerce")
```

```
df["high_cost_flag"] = (df["claim_amount"] >= df["claim_amount"].quantile(0.75)).astype(int)
df.head()
```
