Insurance Claims Cost Analysis notebook.


```python
import pandas as pd
df = pd.read_csv("../data/claims_cost.csv", parse_dates=["claim_date"])
df.head()
```

```python
df.groupby("claim_type")["claim_amount"].agg(["count", "sum", "mean"]).sort_values("sum", ascending=False)
```

```python
df.groupby("region")["claim_amount"].agg(["count", "sum", "mean"]).sort_values("sum", ascending=False)
```
